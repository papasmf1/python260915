```python
from pandas import DataFrame, Series
import pandas as pd

#시리즈 객체 생성
myCampus = Series(data=[100,200,300,400,500], 
index=["2026-09-15", "2026-09-16", "2026-09-17", 
       "2026-09-18", "2026-09-19"])

#슬라이싱
myCampus["2026-09-16"]
#셀단위: shift + enter 
```

```python
#딕셔너리를 데이터프레임으로 변환
data = {"foreigner":[1,2,3,4,5,6],
        "sratio":[10,20,30,40,50,60], 
        "org":[100,200,300,400,500,600], 
        "sprice":[1,2,3,4,5,6], 
        "private":[10,20,30,40,50,60]
        }

df = DataFrame(data)
df.info()  
```

```python
#컬럼을 순서를 지정 
df2 = DataFrame(data=data, 
    columns=["private","org","foreigner","sratio","sprice"],
    index=["2026-09-15","2026-09-16","2026-09-17",
           "2026-09-18","2026-09-19","2026-09-20"])

df2 
```

```python
#특정컬럼만 슬라이싱
df2["org"]
```

```python
#특정행을 슬라이싱: .loc (location)
df2.loc["2026-09-16"]
```

```python
#피봇팅
df2.T 
df2.transpose()
```
